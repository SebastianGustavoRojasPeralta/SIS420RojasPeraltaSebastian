```python
import numpy as np

acciones_posibles=[0, 1, 2] 
alpha=0.5  
turno=1

H=np.zeros(3)  
recompensa_promedio= 0.0 

exponenciales= np.exp(H)
probabilidades= exponenciales / np.sum(exponenciales)
accion_elegida= np.random.choice(acciones_posibles, p=probabilidades)

print(f"turno {turno}")
print(f"Probabilidades del agente: {probabilidades}")
print(f"tomo la accion: {accion_elegida}")

obstaculo_actual= np.random.choice(acciones_posibles) 
print(f"obstculo oculto en la accion: {obstaculo_actual}")

if accion_elegida== obstaculo_actual:
    recompensa= -10  
    print("El agente choco contra el obstaculo recompensa: -10")
else:
    recompensa= 10   
    print("El agente encontro el camino libre recompensa: +10")

for accion in acciones_posibles:
    if accion== accion_elegida:
        H[accion] += alpha * (recompensa - recompensa_promedio) * (1 - probabilidades[accion])
    else:
        H[accion] -= alpha * (recompensa - recompensa_promedio) * probabilidades[accion]

recompensa_promedio += (1 / turno) * (recompensa - recompensa_promedio)

print(f"Memoria actualizada - Nuevas Preferencias (H): {H}")
```

entorno: es la pista por donde avanza. no dibuje paredes, solo hago que aparezca un obstaculo al azar y reparta premios o castigos.

estados: es simplemente el momento exacto en que el agente debe decidir hacia donde dar su siguiente paso.

recompensa: es el puntaje. si el camino esta libre le doy +10, pero si justo pisa el obstaculo le quito 10.

funcion de accion: como use el algoritmo del gradiente, no uso los clasicos puntajes q, sino "preferencias". es una tabla interna que sube cuando el agente hace un buen movimiento.

politica: aplique la formula softmax. esta convierte las preferencias en porcentajes (ej. 80% arriba, 20% abajo) y el agente lanza un dado para decidir su paso en base a eso.

exploracion y explotacion: el gradiente hace ambas a la vez. explota porque el dado casi siempre cae en el camino con mayor porcentaje, pero explora porque ningun camino se queda con 0%, asi que de vez en cuando el dado lo obliga a probar un camino viejo para ver si ya esta libre.
